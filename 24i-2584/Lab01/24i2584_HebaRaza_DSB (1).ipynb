```python
#task1

import pandas as pd
import numpy as np
df = pd.read_csv('adultt.csv')
df.info()
df.head()
print(df.isnull().sum())   
df.replace('?', np.nan, inplace=True)  
print(" ")
print(df.isnull().sum())   
```

```python
#task2

missing_count=df.isnull().sum()
missing_percent= (df.isnull().mean() * 100).round(2)

missing_summary = pd.DataFrame({
    'missing_count': missing_count,
    'missing_percent': missing_percent
}).sort_values(by='missing_count', ascending=False)

print(missing_summary[missing_summary['missing_count'] > 0])
```

```python
#task3

df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
df['native.country'] = df['native.country'].fillna(df['native.country'].mode()[0])

print(df.isnull().sum())

#Workclass & Occupation: Replaced missing values with "Unknown" since assigning a specific category without knowing the true value could introduce incorrect information.
#Native Country: Replaced missing values with the mode, since using the most frequent category avoids losing data while giving a reasonable estimate.
```

```python
#task4
exact_duplicates = df.duplicated().sum()
print(f"Exact duplicate rows: {exact_duplicates}")
duplicates_ignoring_label = df.drop(columns=['income']).duplicated().sum()

print(f"Duplicates ignoring income column: {duplicates_ignoring_label}")
df[df.duplicated(keep=False)]


#If duplicates_ignoring_label is higher than exact_duplicates, that tells you there are rows with identical attributes but different income values — exactly the case the task wants you to reason about and decide not to blindly drop.
```

```python
#task5
print(df['education'].unique())
print(df['marital.status'].unique())
print(df['native.country'].unique())

df['education'] = df['education'].str.strip()
df['marital.status'] = df['marital.status'].str.strip()
df['native.country'] = df['native.country'].str.strip()

print(df['education'].unique())
print(df['marital.status'].unique())
print(df['native.country'].unique())
```

```python
#task6

print(df.describe())
print(df.describe(include='object'))

mean_age = df['age'].mean()
median_age = df['age'].median()
mean_hours = df['hours.per.week'].mean()
median_hours = df['hours.per.week'].median()

print(f"Age -> Mean: {mean_age:.2f}, Median: {median_age}")
print(f"Hours/week -> Mean: {mean_hours:.2f}, Median: {median_hours}")
```

```python
#task7

#(a)
most_common_occupation = df['occupation'].value_counts().idxmax()
print(f"Most common occupation: {most_common_occupation}")
print(df['occupation'].value_counts())

#(b)
sex_distribution = df['sex'].value_counts(normalize=True) * 100
print(sex_distribution.round(2))

#(c)
income_distribution = df['income'].value_counts(normalize=True) * 100
print(income_distribution.round(2))
```

```python
#task8

education_mapping = df.groupby('education')['education.num'].unique()
print(education_mapping)

inconsistent = education_mapping[education_mapping.apply(len) > 1]
print("Inconsistent mappings found:" if len(inconsistent) > 0 else "No inconsistencies found — mapping is consistent.")
print(inconsistent)
```

```python
## Task 10 — Data Profiling Summary Report

**Dataset Overview**
The dataset contains 32,561 rows and 15 columns, representing individual demographic and employment records from the UCI Adult Income dataset. It captures attributes such as age, workclass, education, occupation, and hours worked per week, with the target variable `income` indicating whether a person earns `<=50K` or `>50K` annually.

**Data Quality Issues Found**
- **Missing values:** Hidden as `" ?"` strings rather than proper NaNs. After conversion, `workclass` (~5.6%), `occupation` (~5.7%), and `native.country` (~1.8%) had missing entries; no numeric columns had missing values.
- **Duplicates:** A small number of exact duplicate rows were found and removed. Rows duplicated on all attributes except `income` were also found but kept, since two individuals can share attributes yet have different income outcomes.
- **Inconsistent data:** `education`, `marital.status`, and `native.country` contained leading/trailing whitespace inconsistencies (e.g. `" Bachelors"` vs `"Bachelors"`), which were fixed with `.str.strip()`.

**Key Observations**
- Income classes are **imbalanced**: roughly 75% `<=50K` vs 25% `>50K`.
- The most common occupation is **Prof-specialty** (closely followed by Craft-repair and Exec-managerial).
- The dataset skews male (~67%) vs female (~33%).
- Mean and median for `age` and `hours.per.week` were close, indicating no major skew.
- `education` and `education.num` were verified to be consistently mapped, with no inconsistencies found.

**Cleaning Decisions Made**
- Replaced `" ?"` with `NaN` to expose hidden missing values.
- Imputed `workclass` and `occupation` with `"Unknown"` rather than mode, since missingness there likely reflects real situations (e.g. unemployment) — guessing a specific job could introduce false information.
- Imputed `native.country` with its mode, since missingness appeared more random and one value (`United-States`) dominates the column.
- Dropped exact duplicate rows; kept rows duplicated on all columns except `income`.
- Stripped whitespace from categorical text columns to fix inconsistent category labels.
```
